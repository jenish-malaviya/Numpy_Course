```python
import numpy as np
```

Basic Arithmetic

```python
# 1 Addition
a = np.array([10,20,30,40,50])
b = np.array([60,70,80,90,100])
print(a + b)
```

```python
# 2 Subtraction
a = np.array([10,20,30,40,50])
b = np.array([60,70,80,90,100])
print(a - b)
```

```python
# 3 Divison
a = np.array([10,20,30,40,50])
b = np.array([60,70,80,90,100])
print(a / b)
```

```python
# 4 Multiplication
a = np.array([10,20,30,40,50])
b = np.array([60,70,80,90,100])
print(a * b)
```

```python
# 5 Modulus
a = np.array([10,21,32])
b = np.array([3,4,5])
print(a % b)
```

```python
# 6 Power 
a = np.array([1,2,3,4,5])
print(a ** 2)
```

```python
# 7 Floor Division
a = np.array([152,204,312,441,510])
b = np.array([10,20,30,40,50])
print(a // b)
```

Numpy Functions

```python
# 1 NumPy function for element-wise addition
print(np.add(a,b))
```

```python
# 2 NumPy function for element-wise subtraction.
print(np.subtract(a,b))
```

```python
# 3 NumPy function for element-wise multiplication
print(np.multiply(a,b))
```

```python
# 4 NumPy function for element-wise division
print(np.divide(a,b))
```

```python
# 5 NumPy function for element-wise remainder
print(np.mod(a,b))
```

```python
# 6 Raises each element to a power.
a = np.array([2,3,4])
print(np.power(a,2))
```

```python
# 7 Finds the square root.
a = np.array([4,9,16,25])
print(np.sqrt(a))
```

```python
# 8 Finds the square of each value
a = np.array([4,9,16,25])
print(np.square(a))
```

```python
# 9 Returns absolute values, removing negative signs
a = np.array([-10,-5,0,5,10])
print(np.abs(a))
```

```python
# 10 Returns -1 for negative, 0 for zero, and 1 for positive.
a = np.array([-10,0,25])
print(np.sign(a))
```
