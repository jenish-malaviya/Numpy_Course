```python
import numpy as np 
```

```python
# 1 concatenate Joins arrays along an existing axis. It does not create a new axis
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate((a,b)))
```

```python
# 2 stack Joins arrays along a new axis. The input arrays must have the same shape
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.stack((a,b), axis=1))
```

```python
# 3 vstack Stacks arrays vertically, usually increasing the number of rows.
a = np.array([[1,2,],[3,4]])
b = np.array([[5,6,],[7,8]])

print(np.vstack((a,b)))
```

```python
# 4 hstack Stacks arrays horizontally, usually increasing the number of columns
print(np.hstack((a,b)))
```

```python
# 5 dstack Stacks arrays along the third axis (depth). For 2-D arrays, this creates a 3-D result.
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])

print(np.dstack((a,b)))
```

```python
# 6 column_stack Stacks 1-D arrays as columns. For 2-D arrays, it behaves like horizontal stacking.
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.column_stack((a,b)))
```

```python
# 7 row_stack Stacks 1-D arrays as rows. It is equivalent to vstack() for this common use.
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.row_stack((a,b)))
```

```python
# concatenate = join existing axis
# stack = create a new axis
# vstack = vertical / rows
# hstack = horizontal / columns
# dstack = depth / 3rd axis
# column_stack = columns
# row_stack = rows
```
