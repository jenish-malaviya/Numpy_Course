```python
import numpy as np 
```

```python
# 1 np.random is NumPy's module for random-number generation.
print(np.random.rand())
```

```python
# 2 Generates random floats from 0 (inclusive) to 1 (exclusive).
arr = np.random.rand(3)
print(arr)
```

```python
# For 2 rows and 3 columns
arr2 = np.random.rand(2,3)
print(arr2)
```

```python
# 3 Generates values from the standard normal distribution (mean 0, standard deviation 1).
arr = np.random.randn(5)
print(arr)
```

```python
# 4 Generates random integers; the upper limit is excluded.
arr = np.random.randint(1,10,6)
print(arr)
```

```python
# 5 Generates random floats from 0 (inclusive) to 1 (exclusive).
arr = np.random.random(4)
print(arr)
```

```python
# 6 Generates random floating-point values in a chosen range.
arr = np.random.uniform(10,20,5)
print(arr)
```

```python
# 7 Generates values from a normal distribution. The arguments are mean, standard deviation, and size.
# 50 = mean, 10 = standard deviation, 5 = number of values.
arr = np.random.normal(50,10,5)
print(arr)
```

```python
# 8 Randomly selects values from a sequence.
arr = np.array([10,20,30,40])
x = np.random.choice(arr)
print(x)

# For 2 Values Selections 
np.random.choice(arr,2)
```

```python
# 9 Shuffles the original array in place.
arr = np.array([1,2,3,4,5])
np.random.shuffle(arr)
print(arr)
```

```python
# 10 Returns a shuffled permutation without changing the original array.
arr = np.array([1,2,3,4,5])
new_arr = np.random.permutation(arr)
print(new_arr)
print(arr)
```

```python
# 11 Sets the starting state of NumPy's legacy global random generator, making repeated runs reproducible.
np.random.seed(10)
print(np.random.randint(1,10,5))
```

```python
# 12 Creates the modern NumPy random-number Generator.
rng = np.random.default_rng()
print(rng.random(5))
```

```python
# 13 The object returned by default_rng() is a Generator. Its methods create random values.
rng = np.random.default_rng()
print(rng.integers(1,10,5))
```

```python
# 14 Use the same seed with default_rng() to reproduce a sequence.
rng1 = np.random.default_rng(42)
print(rng1.integers(1,100,5))

rng2 = np.random.default_rng(42)
print(rng2.integers(1,100,5))
```

```python
# rand() → floats 0 to 1
# randn() → standard normal values
# randint() → random integers
# random() → floats 0 to 1
# uniform() → floats in a range
# normal() → normal distribution
# choice() → random selection
# shuffle() → changes original array
# permutation() → returns shuffled permutation
# seed() → reproducibility for legacy global RNG
# default_rng() → modern Generator
```
