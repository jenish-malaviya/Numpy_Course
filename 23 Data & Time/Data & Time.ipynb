```python
import numpy as np
```

```python
# 1 Create a Date
date = np.datetime64("2026-08-23")
print(date)
```

```python
# 2 Create Date and Time
dt = np.datetime64("2026-08-26T18:30")
print(dt)
```

```python
# 3 Date Creation
start = np.datetime64("2026-01-01")
print(start)
```

```python
# 4 Date Arithmetic — Add Days
date = np.datetime64("2026-08-26")
new_date = date + np.timedelta64(5, "D")
print(new_date)
```

```python
# 5 Date Arithmetic — Subtract Days
date = np.datetime64("2026-08-26")
new_date = date - np.timedelta64(10, "D")
print(new_date)
```

```python
# 6 Store a Time Difference
days = np.timedelta64(7, "D")
print(days)
```

```python
# 7 Date Differences
date1 = np.datetime64("2026-08-01")
date2 = np.datetime64("2026-08-10")
difference = date2 - date1
print(difference)
```

```python
# 8 Date Ranges
dates = np.arange(
 np.datetime64("2026-08-01"),
 np.datetime64("2026-08-06")
)
print(dates)
```

```python
# 9 Date Ranges
dates = np.arange(
 np.datetime64("2026-08-01"),
 np.datetime64("2026-08-11"),
 np.timedelta64(2, "D")
)
print(dates)
```

```python
# 10 Business Days
dates = np.arange(
 np.datetime64("2026-08-24"),
 np.datetime64("2026-08-29"),
 dtype="datetime64[D]"
)
print(dates)
print(np.is_busday(dates))
```

```python
# 11 Check a Date
date = np.datetime64("2026-08-29")
print(np.is_busday(date))
```

```python
# 12 Count Business Days
start = np.datetime64("2026-08-24")
end = np.datetime64("2026-08-29")
count = np.busday_count(start, end)
print(count)
```

```python
# 13  np.busday_count() — Simple Example
start = np.datetime64("2026-08-01")
end = np.datetime64("2026-08-10")
print(np.busday_count(start, end))
```

```python
# 14 Date Difference + Business Days
start = np.datetime64("2026-08-24")
end = np.datetime64("2026-09-01")

calendar_days = end - start
business_days = np.busday_count(start, end)

print("Calendar days:", calendar_days)
print("Business days:", business_days)
```

```python
# np.datetime64() -> Creates dates and times
# np.timedelta64() -> Represents a time difference
# date + timedelta -> Adds time
# date - timedelta -> Subtracts time
# date2 - date1 -> Finds date difference
# np.arange() -> Creates date ranges
# np.is_busday() -> Checks business day
# np.busday_count() -> Counts business days
```
