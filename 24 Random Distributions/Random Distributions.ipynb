```python
import numpy as np
```

```python
# 1 Uniform Distribution — np.random.uniform
# Uniform distribution gives random values where every value in the selected range has equal
# chance.
np.random.seed(1)
result = np.random.uniform(10,20,5)
print(result)
```

```python
# 2 Normal Distribution — np.random.normal
# Normal distribution is centered around a mean. The standard deviation controls how spread
# out the values are.
np.random.seed(1)

result = np.random.normal(
    loc=50,
    scale=10,
    size=5
)
print(result)
```

```python
# 3  Binomial Distribution — np.random.binomial
#  Binomial distribution counts the number of successes in a fixed number of independent
# trials.
np.random.seed(1)
result = np.random.binomial(
 n=10,
 p=0.5,
 size=5
)
print(result)
```

```python
# 4 Poisson Distribution — np.random.poisson
# Poisson distribution models how many times an event happens in a fixed interval when an
# average rate is known.
np.random.seed(1)
result = np.random.poisson(
 lam=4,
 size=5
)
print(result)
```

```python
# 5 Exponential Distribution — np.random.exponential
#  Exponential distribution is commonly used for waiting time between events.
np.random.seed(1)
result = np.random.exponential(
 scale=2,
 size=5
)
print(result)
```

```python
# 6 Standard Normal Distribution — np.random.randn
# Standard normal distribution has mean 0 and standard deviation 1.
np.random.seed(1)
result = np.random.randn(5)
print(result)
```

```python
# 7 Mean
# Mean is the average of values. Add all values and divide by the number of values.
data = np.array([10, 20, 30, 40, 50])
result = np.mean(data)
print(result)
```

```python
# 8 Standard Deviation
# Standard deviation tells us how far values are spread from the mean.
data = np.array([10, 20, 30, 40, 50])
result = np.std(data)
print(result)
```

```python
# 9 Probability
# Probability tells us how likely an event is. It ranges from 0 to 1, or 0% to 100%.
probability = 1 / 6
print(probability)
print(probability * 100)
```

```python
# 10 Random Sampling
# Random sampling means selecting observations randomly from a population.
np.random.seed(1)
data = np.array([10, 20, 30, 40, 50])
sample = np.random.choice(data, size=3, replace=False)
print(sample)
```

```python
# 11 Simulation
# Simulation means using random numbers to imitate a real-world process.
np.random.seed(1)
coin_flips = np.random.choice(
 ["Head", "Tail"],
 size=10
)
print(coin_flips)
```

```python
# np.random.uniform() — Random values in a range; equal chance across the range.
# np.random.normal() — Random values based on mean and standard deviation.
# np.random.binomial() — Number of successes in fixed trials.
# np.random.poisson() — Number of events for a given average rate.
# np.random.exponential() — Waiting time between events.
# np.random.randn() — Standard normal values: mean 0, standard deviation 1.
# Mean — Average value.
# Standard deviation — Amount of spread around the mean.
# Probability — Chance that an event happens.
# Random sampling — Selecting observations randomly.
# Simulation — Using random data to imitate a real process.
```
