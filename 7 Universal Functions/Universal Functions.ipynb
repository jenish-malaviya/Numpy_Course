```python
import numpy as np 
```

```python
# 1. What is a ufunc?
# A ufunc means Universal Function. It performs operations efficiently on NumPy array elements.
arr = np.array([1,4,9])
print(np.sqrt(arr))
```

```python
# 2. Element-wise operations
# The operation is performed separately on each element
arr = np.array([1,4,9,16])
print(np.sqrt(arr))
```

```python
# 3. Unary ufunc
# A unary ufunc takes one input
arr = np.array([1,4,9])
result = np.sqrt(arr)
print(result)
```

```python
# 4. Binary ufunc
# A binary ufunc works with two inputs.
a = np.array([10,20,30])
b = np.array([1,2,3])
print(np.add(a,b))
```

```python
# 5. np.sqrt()
# Square root of each element.
a = np.array([4,9,16,25])
print(np.sqrt(a))
```

```python
# 6. np.exp()
# Calculates e raised to each value.
a = np.array([0,1,2])
print(np.exp(a))
```

```python
# 7. np.log()
# Natural logarithm, base e
a = np.array([1,np.e,np.e ** 2])
print(np.log(a))
```

```python
# 8 np.log10
# Base - 10 Logarithm
a = np.array([1,10,100,1000])
print(np.log10(a))
```

```python
# 8.1 np.log2
# Base - 2 Logarithm
a = np.array([1,10,100,1000])
print(np.log2(a))
```

```python
# 9. np.sin()
# Sine of angles in radians.
a = np.array([0,np.pi / 2,np.pi])
print(np.sin(a))
```

```python
# 10. np.cos()
# Cosine of angles in radians. 
a = np.array([0,np.pi / 2,np.pi])
print(np.cos(a))
```

```python
# 11. np.tan()
# Tangent of angles in radians.
a = np.array([0,np.pi / 4])
print(np.tan(a))
```

```python
# 12. np.floor()
# Rounds down to the nearest integer.
a = np.array([1.2,2.9,-1.2])
print(np.floor(a))
```

```python
# 13. np.ceil()
# Rounds up to the nearest integer.
a = np.array([1.2,2.1,-1.2])
print(np.ceil(a))
```

```python
# 14. np.round()
# Rounds to a chosen number of decimal places.
a = np.array([1.234,2.567,3.891])
print(np.round(a,2))
```

```python
# 15. np.trunc()
# Removes the fractional part by moving toward zero
a = np.array([1.9,2.7,-1.9,-2.7])
print(np.trunc(a))
```

```python
# 16. np.absolute()
# Returns absolute values; equivalent to np.abs()
a = np.array([-10,-5,0,5,10])
print(np.absolute(a))
```

```python
# unction	Meaning	Example
# np.sqrt()	Square root	        np.sqrt([4, 9]) → [2, 3]
# np.exp()	Exponential	        np.exp([1, 2])
# np.log()	Natural logarithm	np.log([1, 10])
# np.log10()  Base-10 logarithm	np.log10([10, 100])
# np.sin()	Sine	            np.sin(a)
# np.cos()	Cosine	            np.cos(a)
# np.floor() Round down	        np.floor([1.2, 2.9]) → [1, 2]
# np.ceil()	Round up	        np.ceil([1.2, 2.1]) → [2, 3]
# np.abs()	Absolute value	    np.abs([-5, 3]) → [5, 3]
```
