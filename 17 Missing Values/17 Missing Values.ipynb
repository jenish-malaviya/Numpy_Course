```python
import numpy as np
```

```python
# 1 nan
# nan means Not a Number. It is commonly used to represent a missing value in a NumPy array
arr = np.array([10,20,np.nan,40])
print(arr)
```

```python
# 2 isnan Checks which values are NaN. It returns True for NaN and False for normal numbers.
arr = np.array([10,20,np.nan,30,np.nan])
result = np.isnan(arr)
print(result)
```

```python
# 3 isfinite Checks whether values are finite. Normal numbers are finite; NaN and infinity are not finite.
arr = np.array([10,np.nan,-np.inf,40])
result = np.isfinite(arr)
print(result)
```

```python
# 4 isinf Checks whether values are positive or negative infinity.
arr = np.array([10,np.inf,-np.inf,40])
result = np.isinf(arr)
print(result)
```

```python
# 5 nanmean Calculates the mean while ignoring NaN values.
arr = np.array([10,20,np.nan,40])
result = np.nanmean(arr)
print(result)
```

```python
# 6 nanmedian Calculates the median while ignoring NaN values
arr = np.array([10,20,np.nan,40])
result = np.nanmedian(arr)
print(result)
```

```python
# 7 nanstd Calculates standard deviation while ignoring NaN values.
arr = np.array([10,20,np.nan,40])
result = np.nanstd(arr)
print(result)
```

```python
# 8 nanvar Calculates variance while ignoring NaN values.
arr = np.array([10,20,np.nan,40])
result = np.nanvar(arr)
print(result)
```

```python
# 9 nansum Calculates the sum while ignoring NaN values.
arr = np.array([10,20,np.nan,40,50])
result = np.nansum(arr)
print(result)
```

```python
# 10 nanmin Finds the minimum value while ignoring NaN values.
arr = np.array([10,20,np.nan,40])
result = np.nanmin(arr)
print(result)
```

```python
# 11 nanmax Finds the maximum value while ignoring NaN values.
arr = np.array([10,20,np.nan,40])
result = np.nanmax(arr)
print(result)
```

```python
# 12  Why use nan-functions 
# Normal functions such as np.mean(), np.median(), np.std(), np.var(), np.sum(), np.min(), and np.max() can
# return NaN when the array contains NaN. The nan-functions ignore NaN values and calculate using the
# available numeric values

arr = np.array([10, 20, np.nan, 40])
print(np.mean(arr))
print(np.nanmean(arr))
print(np.sum(arr))
print(np.nansum(arr))
```

```python
#  np.nan → represents a missing/NaN value.
# • np.isnan() → checks for NaN.
# • np.isfinite() → checks for finite values.
# • np.isinf() → checks for infinity.
# • np.nanmean() → mean, ignoring NaN.
# • np.nanmedian() → median, ignoring NaN.
# • np.nanstd() → standard deviation, ignoring NaN.
# • np.nanvar() → variance, ignoring NaN.
# • np.nansum() → sum, ignoring NaN.
# • np.nanmin() → minimum, ignoring NaN.
# • np.nanmax() → maximum, ignoring NaN.
```
