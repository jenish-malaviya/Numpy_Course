```python
import numpy as np
```

```python
arr = np.array([[10,20,30],
               [40,50,60]])
arr
```

```python
arra = np.array([[[10,20,30],
                 [40,50,60],
                 [70,80,90]]])
arra
```

```python
# 1 ndim — Number of Dimensions
print(arr.ndim)
print(arra.ndim)
```

```python
# 2 shape — Size of Each Dimension
print(arr.shape) 
print(arra.shape) 
```

```python
# 3 size — Total Number of Elements
print(arr.size)
print(arra.size)
```

```python
# 4 dtype — Data Type of Elements
print(arr.dtype)
print(arra.dtype)
```

```python
# 5 itemsize — Bytes Used by One Element
print(arr.itemsize)
```

```python
# 6 nbytes — Total Bytes Used by Array Data
print(arr.nbytes)
print(arra.nbytes)
```

```python
# 7 T — Transpose of the Array
print(arr.T)
print(arra.T)
```

```python
# 8  flags — Information About the Array
print(arr.flags)
print(arra.flags)
```

```python
# ndim → Number of dimensions
# shape → Rows and columns / size of each dimension
# size → Total number of elements
# dtype → Data type of elements
# itemsize → Bytes used by one element
# nbytes → Total bytes used by array data
# T → Transpose
# flags → Array storage/access information
```
