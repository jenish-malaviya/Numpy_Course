```python
import numpy as np
```

```python
# 1 Matrix Basic A matrix is a rectangular arrangement of numbers in rows and columns
a = np.array([[1,2],
              [3,4]])

print(a)
```

```python
# 2 Matrix Creation Create a matrix using np.array().
a = np.array([[10,20,30],
              [40,50,60]])

print(a)
print("shape:",a.shape)
```

```python
# 3 Matrix Addition Add corresponding elements of two matrices.
a = np.array([[1,2],
              [3,4]])

b = np.array([[5,6],
              [7,8]])

print(a + b)
```

```python
# 4 Matrix Subtraction Subtract corresponding elements.
A = np.array([[10, 20],
              [30, 40]])

B = np.array([[1, 2],
              [3, 4]])

print(A - B)
```

```python
# 5 Matrix Multiplication Matrix multiplication uses rows of the first matrix and columns of the second matrix.
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
print(A @ B)

# (1×5)+(2×7)
# =5+14=
# 19
# 	​
# (1×6)+(2×8)
# =6+16=
# 22

# (3×5)+(4×7)
# =15+28=
# 43
# 	​
# (3×6)+(4×8)
# =18+32=
# 50
	​
```

```python
# 6 Element-wise Multiplication Multiply elements at the same positions.
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
print(A * B)
```

```python
# 7 Transpose Transpose changes rows into columns.
a = np.array([[1,2,3],
              [4,5,6]])

print(a.T)
```

```python
# 8  Identity Matrix An identity matrix has 1s on the main diagonal and 0s elsewhere.
I = np.eye(3)

print(I)
```

```python
# 9 Diagonal Matrix A diagonal matrix has values on the main diagonal and zeros elsewhere.
D = np.diag([10,20,30])

print(D)
```

```python
# 10  np.dot np.dot() performs a dot product. For 2-D matrices, it can perform matrix multiplication.
a = np.array([[1,2],
              [3,4]])

b = np.array([[5,6],
              [7,8]])

print(np.dot(a,b))
```

```python
# 11 matmul performs matrix multiplication. 
a = np.array([[1,2],
              [3,4]])

b = np.array([[5,6],
              [7,8]])

print(np.matmul(a,b))
```

```python
# 12 linalg.det Finds the determinant of a square matrix.
a = np.array([[1,2],
              [3,4]])

print(np.linalg.det(a))

# (1×4)−(2×3)
# 1×4=4
# 2×3=6

# 4−6=−2
```

```python
# 13 linalg.inv Finds the inverse of an invertible square matrix.
a = np.array([[1,2],
              [3,4]])

print(np.linalg.inv(a))
```

```python
# 14 linalg.solve Solves a system of linear equations. Here: x + y = 5 and 2x + 3y = 13.
a = np.array([[1,1],
              [2,3]])

b = np.array([5,13])

x = np.linalg.solve(a,b)
print(x)
```

```python
# 15 inalg.eig Finds eigenvalues and eigenvectors
A = np.array([[2, 0],
 [0, 3]])
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:")
print(vectors)
```

```python
# 16 linalg.eigvals Finds only the eigenvalues.
A = np.array([[2, 0],
 [0, 3]])
print(np.linalg.eigvals(A))
```

```python
# 17 linalg.svd SVD means Singular Value Decomposition. It decomposes a matrix into U, singular values, and Vt.
A = np.array([[1, 0],
 [0, 2]])
U, S, Vt = np.linalg.svd(A)

print("U:")
print(U)
print("Singular values:")
print(S)
print("Vt:")
print(Vt)
```
