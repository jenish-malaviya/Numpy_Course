```python
import numpy as np
```

```python
arr = np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])
arr
```

```python
# 1 Row Selection
print(arr[1])
```

```python
# 2 Column Selection
print(arr[:,1])
```

```python
# 3 Specifi Element
print(arr[1,2])
```

```python
# 4 Multiple Rows
print(arr[[0,2]])
```

```python
# 5 Multiple Rows
print(arr[:,[0,2]])
```

```python
# 6 Row Slicing
print(arr[0:2, :])
```

```python
# 7 Column Slicing
print(arr[:,0:2])
```
