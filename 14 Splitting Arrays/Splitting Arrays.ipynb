```python
import numpy as np
```

```python
# 1 split normally requires equal-sized parts.
arr = np.array([10,20,30,40,50,60])
result = np.split(arr,3)
print(result)
```

```python
# 2 array_split can split an array into unequal parts when necessary.
arr = np.array([10,20,30,40,50])
result = np.array_split(arr,3)
print(result)
```

```python
# 3 hsplit Splits columns horizontally.
arr = np.array([[10,20,30,40],
                [50,60,70,80]])

result = np.hsplit(arr,2)
print(result)
```

```python
# 4 vsplit Splits rows vertically
arr = np.array([[10,20],
                [30,40],
                [50,60],
                [70,80]])

result = np.vsplit(arr,2)
print(result)
```

```python
# 5 dsplit Splits the third/depth dimension of a 3D array
arr = np.array([[[10,20],
                 [30,40]],
                [[50,60],
                 [70,80]]])
result = np.dsplit(arr,2)
print(result)
```
